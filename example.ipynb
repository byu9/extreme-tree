```python
import matplotlib.pyplot as plt
import pandas as pd
from scipy.stats import genextreme

from extreme_tree import ExtremeForest

synthetic_test_case = {
    # (x1, x2): (mu, sigma, xi)
    (0, 0): (10.0, 1.0, 0.00),
    (0, 1): (15.0, 1.5, 0.10),
    (1, 0): (20.0, 2.0, -0.10),
    (1, 1): (25.0, 3.0, 0.20),
    (2, 0): (30.0, 3.5, -0.20),
    (0, 2): (18.0, 2.0, 0.15),
    (2, 2): (35.0, 4.0, 0.30),
    (-1, -1): (8.0, 0.8, -0.25),
}


synthetic_dataset = pd.concat(
    [
        pd.DataFrame(
            {
                "x1": x1,
                "x2": x2,
                "mu": mu,
                "sigma": sigma,
                "xi": xi,
                "y": genextreme.rvs(c=-xi, loc=mu, scale=sigma, size=1000),  # scipy uses c = -xi
            }
        )
        for (x1, x2), (mu, sigma, xi) in synthetic_test_case.items()
    ],
    ignore_index=True,
)

synthetic_dataset
```

```python


model = ExtremeForest(
    ensemble_size=50,
    resample_ratio=1.0,
    max_n_splits=40,
    min_partition_size=50,
    min_score=0.01,
)


feature = synthetic_dataset[["x1", "x2"]]
target = synthetic_dataset["y"]


model.fit(feature, target)
mu_hat, sigma_hat, xi_hat = model.predict(feature)
```

```python
# The red line shows the true parameter value.
# The black lines each corresponds to a member estimate.

plt.subplot(311)
plt.plot(mu_hat, color="black", alpha=0.2)
plt.plot(synthetic_dataset["mu"], color="red")
plt.title(r"$\hat{\mu}$ vs $\mu$")
plt.grid()

plt.subplot(312)
plt.plot(sigma_hat, color="black", alpha=0.2)
plt.plot(synthetic_dataset["sigma"], color="red")
plt.title(r"$\hat{\sigma}$ vs $\sigma$")
plt.grid()

plt.subplot(313)
plt.plot(xi_hat, color="black", alpha=0.2)
plt.plot(synthetic_dataset["xi"], color="red")
plt.title(r"$\hat{\xi}$ vs $\xi$")
plt.grid()

plt.gcf().subplots_adjust(hspace=0.8)
```
